```python
import numpy as np
import pandas as pd
import requests # handles HTTP requests
import json # handles JSON data
```

```python
url = 'https://www.thecocktaildb.com/api/json/v1/1/filter.php?c=Cocktail'
```

```python
r = requests.get(url)
r
```

```python
r.text 
```

```python
#r.json()
myjson = json.loads(r.text)
myjson
```

```python
myjson['drinks'][0]['strDrink']
```

**List comprehension:** a loop over item in a list, outputs results to a different list.

```python
[x['strDrink'] for x in myjson['drinks']]
```

Record path, the path to a list of rows in the data frame I am trying to build

In this case, the record path is just: 'drinks'

```python
drinksDF = pd.json_normalize(myjson, record_path=['drinks'])
drinksDF
```
